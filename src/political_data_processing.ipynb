```python
import numpy as np
import pandas as pd
```

```python
abortion = pd.read_csv("abortion.csv", skiprows = [0,1,2,3], delimiter = ',')

abortion.columns = ['Year: Abortion...?', 'Legal in all/most cases (%)', 'Illegal in all/most cases (%)']
abortion.loc[27, 'Year: Abortion...?'] = '2022'
abortion = abortion.drop(abortion.index[30:34]) # no data for 2025
abortion = abortion.drop(abortion.index[0:19])
abortion = abortion.drop([26]) # choosing one for 2022 (the higher value)
abortion = abortion.sort_values(by = "Year: Abortion...?", ascending = False)
abortion['Illegal in all/most cases (%)'] = abortion['Illegal in all/most cases (%)'].replace('%', '', regex = True)
abortion['Legal in all/most cases (%)'] = abortion['Legal in all/most cases (%)'].replace('%', '', regex = True)
abortion = abortion.astype({"Year: Abortion...?": "int64", "Legal in all/most cases (%)": "int64", "Illegal in all/most cases (%)": "int64"})
abortion4project = abortion.set_index('Year: Abortion...?')

# abortion4project.to_csv("abortion4project.csv")
print(abortion4project)
```

```python
trust_gov = pd.read_csv("trust_gov.csv", skiprows = [0,1,2,3], delimiter = ',', header=None)

trust_gov = trust_gov.drop([1,2], axis=1) # choosing the smoothed trend
trust_gov = trust_gov.drop(trust_gov.index[12:171])
trust_gov = trust_gov.drop([7]) # choosing one for 2020 (the higher one)
trust_gov = trust_gov.drop([1]) # choosing one for 2025 (the latest one)
trust_gov = trust_gov.drop([9]) # choosing one for 2017 (the higher one)
trust_gov.columns = ['Year: Government...?', 'Do what is right just about always/most of the time (%)']
trust_gov["Year: Government...?"] = pd.to_datetime(trust_gov["Year: Government...?"]).dt.strftime("%Y")
trust_gov = trust_gov.astype({'Year: Government...?': 'int64', 'Do what is right just about always/most of the time (%)': 'int64'})
trust_gov4project = trust_gov.set_index('Year: Government...?')

# trust_gov4project.to_csv("trust_gov4project.csv")
print(trust_gov4project)
```

```python
corruption = pd.read_csv("corruption.csv") 

corruption = corruption.rename(columns = {"Time": "Year: Government...?", "Yes": "Corrupted (%)"})
corruption = corruption.drop([0])
corruption = corruption.drop(corruption.index[11:21])
corruption['Corrupted (%)'] = corruption['Corrupted (%)'].replace('%', '', regex = True)
corruption = corruption.astype({"Corrupted (%)": "int64"})
corruption4project = corruption.set_index('Year: Government...?')

# corruption.to_csv("corruption4project.csv")
print(corruption4project)
```

```python
economy_conf = pd.read_csv("economy_conf.csv")

economy_conf = economy_conf.rename(columns = {"Unnamed: 0": "Year: Economy...?", "Excellent": "Excellent (%)",\
                                              "Good": "Good (%)", "Only fair": "Only Fair (%)", "Poor": "Poor (%)", "No opinion": "No opinion (%)"})
economy_conf = economy_conf.drop(economy_conf.index[123:335])
economy_conf = economy_conf.dropna()
economy_conf = economy_conf.drop('Excellent (%)', axis=1)
economy_conf = economy_conf.drop('Only Fair (%)', axis=1)
economy_conf = economy_conf.drop('No opinion (%)', axis=1)
economy_conf = economy_conf.astype({"Good (%)": "int64", "Poor (%)": "int64"})
economy_conf["Year: Economy...?"] = pd.to_datetime(economy_conf["Year: Economy...?"].str.extract(r"(\d{4} \w+)")[0],
                                                  format = "%Y %b").dt.strftime("%Y-%m")
economy_conf = economy_conf.drop(economy_conf.index[0:8])
economy_conf4project = economy_conf.set_index('Year: Economy...?')

# economy_conf4project.to_csv("economy_conf4project.csv")
print(economy_conf)
```

```python
economy_job = pd.read_csv("economy_job.csv")

economy_job = economy_job.rename(columns = {"X.1": "Year: Job Market..?", "Good time": "Good Time (%)",\
                                              "Bad time": "Bad Time (%)", "No opinion": "No opinion (%)"})
economy_job = economy_job.drop(economy_job.index[86:255])
economy_job = economy_job.dropna()
economy_job = economy_job.drop(economy_job.index[0:3])
economy_job = economy_job.drop('No opinion (%)', axis = 1)
economy_job = economy_job.astype({"Good Time (%)": "int64", "Bad Time (%)": "int64"})
economy_job["Year: Job Market..?"] = pd.to_datetime(economy_job["Year: Job Market..?"].str.extract(r"(\d{4} \w+)")[0],
                                                  format = "%Y %b").dt.strftime("%Y-%m")
economy_job4project = economy_job.set_index('Year: Job Market..?')

# economy_job4project.to_csv("economy_job4project.csv")
print(economy_job)
```

```python
lgbtq = pd.read_csv("lgbtq.csv")

lgbtq = lgbtq.drop([0])
lgbtq = lgbtq.drop(lgbtq.index[11:27])
lgbtq = lgbtq.rename(columns = {"Do you think marriages between same-sex couples should or should not be recognized by the law as valid, with \nthe same rights as traditional marriages?": "Year: Gay people...?",
                                "% Same-sex marriages should be valid": "Gay marriages valid (%)", "% Gay or lesbian relations are morally acceptable": "Gay couples morally acceptable (%)",
                                "% Changing one's gender is morally acceptable": "Trans people morally acceptable (%)"})
lgbtq = lgbtq.drop('Gay couples morally acceptable (%)', axis=1)
lgbtq['Trans people morally acceptable (%)'] = lgbtq['Trans people morally acceptable (%)'].fillna('0')
lgbtq = lgbtq.astype({'Gay marriages valid (%)': 'int64', 'Trans people morally acceptable (%)': 'int64'})
lgbtq4project = lgbtq.set_index('Year: Gay people...?')

# lgbtq4project.to_csv("lgbtq4project.csv")
print(lgbtq4project)
```

```python
political_affil = pd.read_csv("political_affil.csv")

political_affil = political_affil.dropna()
political_affil = political_affil.rename(columns = {"X.1": "Year: Political Affiliation...?", "Republican/Republican leaner": "Republican/Republican leaning (%)",
                                                    "Democrat/Democratic leaner": "Democrat/Democratic leaning (%)"})
political_affil = political_affil.drop(political_affil.index[0:24])
political_affil = political_affil.sort_values(by = "Year: Political Affiliation...?", ascending = False)
political_affil = political_affil.astype({'Year: Political Affiliation...?': 'int64', "Republican/Republican leaning (%)": "int64", "Democrat/Democratic leaning (%)": "int64"})
political_affil4project = political_affil.set_index('Year: Political Affiliation...?')

# political_affil4project.to_csv("political_affil4project.csv")
print(political_affil4project)
```

```python
vaccines = pd.read_csv("vaccines.csv")

vaccines = vaccines.rename(columns = {"X.1": "Year: Vaccines...?", "% Yes, require": "Mandatory vaccines (%)",
                                      "% No, government should stay out of": "No mandatory vaccines (%)"})
vaccines = vaccines.drop([0])
vaccines = vaccines.sort_values(by = "Year: Vaccines...?", ascending = False)
vaccines4project = vaccines.set_index('Year: Vaccines...?')

# vaccines4project.to_csv("vaccines4project.csv")
print(vaccines4project)
```

```python
immigration = pd.read_csv("immigration.csv")

immigration = immigration.dropna()
immigration = immigration.rename(columns = {"X.1": "Year: Immigration...?", "Good thing": "Good thing (%)", "Bad thing": "Bad thing (%)",
                                      "Mixed (vol.)": "Mixed (%)", "No opinion": "No opinion (%)"})
immigration = immigration.drop(immigration.index[12:26])
immigration = immigration.drop([1])
immigration = immigration.drop('Mixed (%)', axis=1)
immigration = immigration.drop('No opinion (%)', axis=1)
immigration["Year: Immigration...?"] = pd.to_datetime(immigration["Year: Immigration...?"].str.extract(r"(\d{4} \w+)")[0],
                                                  format = "%Y %b").dt.strftime("%Y")
immigration = immigration.astype({"Year: Immigration...?": "int64", "Good thing (%)": "int64", "Bad thing (%)": "int64"})
immigration4project = immigration.set_index('Year: Immigration...?')

# immigration4project.to_csv("immigration4project.csv")
print(immigration4project)
```
